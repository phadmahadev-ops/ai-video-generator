```python
from h3_prompt_enhancer import main, load_sysprompt, call_api
import sys

RAW_PROMPT = "A woman walking down a neon corridor, camera tracking back, no background music"

brief = call_api(
    os.environ['H3_API_URL'], os.environ['H3_API_KEY'], os.environ['H3_MODEL'],
    [{'role': 'system', 'content': load_sysprompt()},
     {'role': 'user', 'content': RAW_PROMPT}])
print(brief)
with open('/content/brief.txt','w') as f: f.write(brief)
```
